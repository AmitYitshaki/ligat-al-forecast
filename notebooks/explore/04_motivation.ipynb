<div dir="rtl" style="text-align:right;line-height:1.75"><h2>2.3 — למה CRN?</h2><p>יחס SE של r אומר שבלי המספרים האקראיים המשותפים נדרשים בקירוב פי r² מסלולים לאותו דיוק: בחציון פי 1.6 לאליפות, פי 2.1 לעליון ופי 1.4 להישארות.</p><p>Common Random Numbers — מספרים אקראיים משותפים. מריצים המשך עונה אחרי ניצחון ואחרי הפסד עם אותם מספרים אקראיים בכל מסלול. למשל, באותו מסלול משחק אחר בין בית״ר לאשדוד יקבל אותו מספר אקראי בשני התרחישים. אם המשתתפות והסיכויים זהים, גם התוצאה המדומה זהה; לפני הפיצול המשתתפות העתידיות יכולות להשתנות, ולכן אין הבטחה שכל שאר הליגה תסתיים בדיוק באותה תוצאה. כך מפחיתים רעש ומבודדים טוב יותר את שינוי תוצאת היום. כדי להשוות נעליים משתמשים באותו רץ ובאותם תנאים. השיטה אינה מתקנת את הנחת Elo המוקפא.</p><p><b>השאלה:</b> כמה רעש היינו מקבלים אילו הגרלנו שני עתידים בלתי תלויים?</p><p><b>מה התא עושה:</b> משווה את שגיאת התקן הנאיבית ל-mc_se המזווג בכל השורות, לכל יעד.</p><p><b>מה רואים / איך לקרוא:</b> יחס מעל 1 אומר שהשגיאה הבלתי תלויה גדולה יותר. מוצגים חציון ורבעונים אמפיריים.</p><p><b>הסבר נתונים:</b> 0/0 אינו יחס ומוחרג; שגיאה מזווגת 0 עם שגיאה נאיבית חיובית נותנת ∞ ונשמרת. n_ratio הוא מספר היחסים המוגדרים.</p><p><b>הסבר כדורגלני:</b> אחרת רעש של שאר המחזורים היה מסתיר את ההשפעה של משחק היום. CRN משווה את אותה ליגה נגד-עובדתית.</p></div>

$$
SE_{paired}=\frac{s(D_s)}{\sqrt N},\quad D_s=I_{T,s}(W)-I_{T,s}(L),\qquad SE_{ind}=\sqrt{\frac{p_W(1-p_W)}N+\frac{p_L(1-p_L)}N}
$$

<div dir="rtl"><p><b>מקרא:</b> s(D) היא סטיית תקן מדגמית, ddof=1; D יכול להיות −1,0,1. שגיאת Monte Carlo שונה מ-SE העונתי בפרק 4.</p></div>

$$
\widehat\Delta_B=\frac{\sum_{h\in H_B}n_{Bh}(\bar r_{Bh}-\bar r_{C,h})}{\sum_{h\in H_B}n_{Bh}}
$$

<div dir="rtl">B קטגוריה; C קבוצת ההשוואה שהוגדרה בסעיף; h שכבה; H_B שכבות שיש בהן שני הצדדים; r בפועל פחות צפוי.</div>

<div dir="rtl" style="text-align:right;line-height:1.75"><h2>4.3 — שיפועים ברמת המשחק</h2><p>שיפוע גדול אינו אומר אפקט גדול: כופלים בפער שקורה בפועל. הטבלה המלאה: 6ב.2. הפער הטיפוסי קבוע במדגם; גם מקדמי כל דגימת bootstrap מוכפלים באותו פער, וכך מתקבל SE באותן יחידות.</p><p><b>השאלה:</b> מה פער החשיבות מוסיף כשהדגל כבר בפנים?</p><p>שורה אחת של הבית לכל משחק. קודם L2/L3/L5/Lc לבדם בכל המשחקים; אחר כך Lc רק במשחקים חיים. המודל המשותף ממרכז את y, L2_diff ו-Lc_diff בתוך אותה שכבה ומבצע OLS עם שני מקדמים וללא חותך. בכל bootstrap ממוצעי השכבות מחושבים מחדש.</p><p>החציון ו-P90 של |diff| מחושבים רק במשחקים שבהם diff אינו 0 ובאוכלוסייה של אותו שיפוע. האפקט בנקודות הוא המקדם כפול הפער. ב-L2 הפער הוא 1 כשצד אחד מת; בשיפוע משותף מדובר בהחזקת המשתנה השני קבוע.</p></div>

$$
\widetilde y_i=b_D\widetilde{L2diff}_i+b_L\widetilde{Lcdiff}_i+\epsilon_i,\quad\widetilde z_i=z_i-\bar z_{h(i)},\quad\widehat{\mathbf b}=(\widetilde X^\top\widetilde X)^+\widetilde X^\top\widetilde y
$$

<div dir="rtl">y שארית נקודות הבית; h השכבה; D הדגל, L החשיבות הרציפה; + הוא פסאודו-הופכי. היחידות הן נקודות ליחידת פער. ליד המשותף מוצגים שני המשתנים לבדם על אותן שורות.</div>

$$

<div dir="rtl" style="text-align:right;line-height:1.75"><h2>2.3 — למה CRN?</h2><p><b>השאלה:</b> כמה רעש היינו מקבלים אילו הגרלנו שני עתידים בלתי תלויים?</p><p><b>מה התא עושה:</b> משווה את שגיאת התקן הנאיבית ל-mc_se המזווג בכל השורות, לכל יעד.</p><p><b>מה רואים / איך לקרוא:</b> יחס מעל 1 אומר שהשגיאה הבלתי תלויה גדולה יותר. מוצגים חציון ורבעונים אמפיריים.</p><p><b>הסבר נתונים:</b> 0/0 אינו יחס ומוחרג; שגיאה מזווגת 0 עם שגיאה נאיבית חיובית נותנת ∞ ונשמרת. n_ratio הוא מספר היחסים המוגדרים.</p><p><b>הסבר כדורגלני:</b> אחרת רעש של שאר המחזורים היה מסתיר את ההשפעה של משחק היום. CRN משווה את אותה ליגה נגד-עובדתית.</p></div>

$$
SE_{paired}=\frac{s(D_s)}{\sqrt N},\quad D_s=I_{T,s}(W)-I_{T,s}(L),\qquad SE_{ind}=\sqrt{\frac{p_W(1-p_W)}N+\frac{p_L(1-p_L)}N}
$$

<div dir="rtl"><p><b>מקרא:</b> s(D) היא סטיית תקן מדגמית, ddof=1; D יכול להיות −1,0,1. שגיאת Monte Carlo שונה מ-SE העונתי בפרק 4.</p></div>

```python
crn = mr.naive_paired_se_comparison(d)
display(
    mv.table(
        crn,
        names=review.names,
        columns=[
            "target",
            "q25",
            "median",
            "q75",
            "n_ratio",
            "infinite_ratio",
            "undefined_ratio",
            "n",
            "n_games",
            "n_seasons",
        ],
    )
)
```

<b>התשובה בקצרה: כנראה שלא. ההכרעה הסופית בניסוי EXP-004 (חלק B).</b></p>
<ol>
<li><b>הדגל ״אין על מה לשחק״ מבדיל בתיאור העבר.</b> אחרי הפחתת ציפיית Elo, הפער בין קבוצות בלי יעד לקבוצות ההשוואה החיות באותו שלב ובאותה יתרת משחקים הוא כ־<span data-claim="flag_gap">0.36</span> נקודות למשחק; זהו פער בין קבוצות, ולא הפסד המתות עצמן מול התחזית (4.1). המתות עצמן צוברות כ־<span data-claim="dead_own">0.15</span> נקודות למשחק פחות מ-Elo, ובחמש העונות האחרונות כ־<span data-claim="dead_recent_own">0.02</span> בלבד; את רוב הפער יוצרות קבוצות ההשוואה, שרובן נאבקות על יעד במחזורי הסיום (4.1.1).</li>
<li><b>מעבר לדגל אין יתרון נקודות ברור.</b> בין קבוצות שעוד נאבקות, מדרגות החשיבות, החשיבות הרציפה והרכיבים לכל יעד אינם מראים יתרון ברור ביחס לאי־הוודאות; החשיבות הרציפה אינה ברורה גם כשהדגל כבר בפנים (4.2, 4.3).</li>
<li><b>גם הדגל מציג בסיס חלש לצפות לשיפור תחזית.</b> הוא נוגע לכ־<span data-claim="one_dead_pct">3.6</span>% מהמשחקים, שבהם רק צד אחד בלי יעד; בחמש העונות האחרונות פער השאריות של הצד המת במשחקים אלה מול קבוצות ההשוואה הוא <span data-claim="recent_flag">−0.20</span> נקודות ואי־הוודאות גדולה מהאפקט. הדגל חזק בעשור הראשון, חלש ולא ודאי בחמש העונות האחרונות; לכן הציפייה שנרשמה מראש היא שאף ייצוג לא יעבור את כלל הכניסה ל-EXP-004 (4.4, פרק הגשר).</li>
<li><b>הממצא הנוסף הוא באופי המשחק.</b> במאבק הישארות צמוד יש כ־<span data-claim="fewer_goals">0.18</span> שערים פחות למשחק מהצפוי לשני הצדדים יחד. בין שתי קבוצות בלי יעד, שיעור התיקו נמוך בכ־<span data-claim="fewer_draw_pp">12</span> נקודות אחוז מהצפוי; בחמש העונות האחרונות החסר קטן יותר. בחלוקה לפי סוג המשחק, בחמש האחרונות יש כ־<span data-claim="recent_both_goals">0.8</span> שערים מעל הצפוי כששתיהן מתות, וכ־<span data-claim="recent_one_goals">1.1</span> כשצד אחד מת ואין מאבק הישארות. במשחקי הפלייאוף הנקיים אין עודף ברור באותן עונות; בתחתון המדגם קטן. לכן אין כאן הסבר מספק של השלב לעודף במשחקי המתות בעונות האחרונות (4.6, 4.6.1). <b>מה זה אומר:</b> לצד האות החלש בנקודות בחמש האחרונות, הדגל עשוי להיות קשור למספר השערים. השערים ייבדקו במחברת 05 (M2), והתיקו במחברת 06 (M3).</li>
<li><b>מה זה אומר למודל:</b> הדגל נשאר קו הבסיס הפשוט להשוואה. החשיבות המחושבת בסימולטור לא הראתה בחלק A ערך מוסף ברור לחיזוי התוצאה; חלק B יבדוק זאת מחוץ למדגם (4.5, פרק הגשר).</li>
</ol>
<p><b>חלק A תיאורי: הוא מראה קשרים בעבר, ולא בודק חיזוי. המבחן החיזויי הוא EXP-004.</b></p>
